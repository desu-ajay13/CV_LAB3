```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
import cv2
r_min=100
r_max=200
img=cv2.imread("/content/drive/MyDrive/cv img/messi.jpg")
```

```python
slicing_with_bg=img.copy()
slicing_without_bg=np.zeros(img.shape,dtype="uint8")
for i in range(img.shape[0]):
  for j in range(img.shape[1]):
    pixel_value=img[i][j]
    if np.all((pixel_value >= r_min) & (pixel_value <= r_max)):
      # If condition is met, set pixel to white in both images
      slicing_with_bg[i][j] = [255, 255, 255] # Set to white
      slicing_without_bg[i][j] = [255, 255, 255] # Set to white
    else:
      slicing_without_bg[i][j] = [0, 0, 0] # Set to black
```

```python
slicing_with_bg=img.copy()
slicing_without_bg=np.zeros(img.shape,dtype="uint8")
for i in range(img.shape[0]):
  for j in range(img.shape[1]):
    pixel_value=img[i][j]
    # Check if all color channels of the pixel are within the defined range
    if np.all((pixel_value >= r_min) & (pixel_value <= r_max)):
      # If condition is met, set pixel to white in both images
      slicing_with_bg[i][j] = [225, 225, 225]
      slicing_without_bg[i][j] = [225, 225, 225]
    else:

      slicing_without_bg[i][j] = [0, 0, 0]
```

```python
def show(img_orig, img_with_bg, img_without_bg):
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img_orig, cmap='gray')
    plt.title('Original Grayscale Image')
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.imshow(img_with_bg, cmap='gray')
    plt.title('Sliced with Background (Grayscale)')
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.imshow(img_without_bg, cmap='gray')
    plt.title('Sliced without Background (Grayscale)')
    plt.axis('off')

    plt.show()


gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

slicing_with_bg = gray_img.copy()
slicing_without_bg = np.zeros(gray_img.shape, dtype="uint8")

for i in range(gray_img.shape[0]):
  for j in range(gray_img.shape[1]):
    pixel_value = gray_img[i, j]

    if r_min <= pixel_value <= r_max:

      slicing_with_bg[i, j] = 255
      slicing_without_bg[i, j] = 255
    else:

      slicing_without_bg[i, j] = 0

show(gray_img, slicing_with_bg, slicing_without_bg)
```
